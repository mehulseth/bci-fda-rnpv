```python
import requests

url = "https://clinicaltrials.gov/api/v2/studies"
params = {
    "query.cond": "epilepsy",
    "filter.advanced": "AREA[InterventionType]DEVICE",
    "pageSize": 100,
    "format": "json",
}
r = requests.get(url, params=params).json()
len(r["studies"])
```

```python
r["studies"][0]["protocolSection"].keys()
```

```python
s = r["studies"][0]["protocolSection"]

print("NCT ID:     ", s["identificationModule"]["nctId"])
print("Title:      ", s["identificationModule"]["briefTitle"])
print("Sponsor:    ", s["sponsorCollaboratorsModule"]["leadSponsor"]["name"])
print("Status:     ", s["statusModule"]["overallStatus"])
print("Start:      ", s["statusModule"].get("startDateStruct", {}).get("date"))
print("Completion: ", s["statusModule"].get("primaryCompletionDateStruct", {}).get("date"))
print("Conditions: ", s["conditionsModule"].get("conditions"))
print("Enrollment: ", s["designModule"].get("enrollmentInfo", {}).get("count"))
print("Allocation: ", s["designModule"].get("designInfo", {}).get("allocation"))
print("Interventions:", [(i["type"], i["name"]) for i in s["armsInterventionsModule"].get("interventions", [])])
print("Sites:      ", len(s.get("contactsLocationsModule", {}).get("locations", [])))
print("FDA-regulated device:", s.get("oversightModule", {}).get("isFdaRegulatedDevice"))
```
